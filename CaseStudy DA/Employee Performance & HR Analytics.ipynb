```python
numerical_cols = data.select_dtypes(include=np.number).columns
#it stores the numerical columns from the dataset
```

```python
categorical_cols = data.select_dtypes(include="object").columns
#it stores the categorical columns from the dataset
```

```python
for col in categorical_cols:
    print(f"\nValue counts for {col}:")
    print(data[col].value_counts())
#checking value counts per categorical column
```

```python
data[numerical_cols].hist(figsize=(16, 12), bins=20, edgecolor="black")
plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
#numerical column distribution
```

```python
department_performance = (data.groupby("Department")["Performance_Score"].mean().sort_values(ascending=False))
print(department_performance)
#average performance score by each department
```

```python
department_performance.plot(kind="bar", figsize=(9, 5), edgecolor="black")
plt.title("Average Performance Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Performance Score")
plt.xticks(rotation=360)
#plt.tight_layout()
plt.show()
#chart for average department performance
```

```python
salary_by_department = (data.groupby("Department")["Salary(LPA)"].agg(["mean", "median", "min", "max"]).round(2))
print(salary_by_department)
#calculate average of min, max, median and mean salary per department
```

```python
sns.barplot(data= data, x="Department", y="Salary(LPA)")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary (LPA)")
plt.xticks(rotation=360)
plt.tight_layout()
plt.show()
# Average salary chart
```

```python
experience_performance = (data.groupby("Years_at_Company")["Performance_Score"].mean())
print(experience_performance)
#average performance by years at company
```

```python
experience_performance.plot(kind="line", marker="o", figsize=(9, 5))
plt.title("Years at Company vs Average Performance")
plt.xlabel("Years at Company")
plt.ylabel("Average Performance Score")
plt.grid(True)
plt.tight_layout()
plt.show()
#line chart to show employees years at company vs their average performance
```

```python
attendance_corr = data[["Attendance_%", "Performance_Score"]].corr().iloc[0, 1]
print("Attendance-Performance Correlation:",round(attendance_corr, 3))
#Correlation between attendance and performance
```

```python
sns.scatterplot(data=data, x="Attendance_%", y="Performance_Score", hue="Department")
plt.title("Attendance vs Performance Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Performance Score")
plt.tight_layout()
plt.show()
#scatterplot to show correlation for attendance vs performance score among different departments
```

```python
correlation_matrix = data[numerical_cols].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap - Employee Performance")
plt.tight_layout()
plt.show()
#heat map to show correlation among numerical columns
```

```python
outlier_cols = [
    "Age",
    "Years_at_Company",
    "Attendance_%",
    "Salary(LPA)",
    "Performance_Score"
]

fig, axes = plt.subplots(1, len(outlier_cols), figsize=(18, 5))

for ax, col in zip(axes, outlier_cols):
    sns.boxplot(y=data[col], ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
#boxplot to show outliers
```

```python
employee_cols = [
    "Employee_ID",
    "Department",
    "Years_at_Company",
    "Goals_Achieved",
    "Projects_Completed",
    "Performance_Score"
]

top_performers = data.nlargest(10, "Performance_Score")
print("Top 10 Performers:")
print(top_performers[employee_cols].to_string(index=False))
#Top 10 employees by performance score

lowest_performers = data.nsmallest(10, "Performance_Score")
print("\nBottom 10 Performers:")
print(lowest_performers[employee_cols].to_string(index=False))
#Bottom 10 employees by performance score
```

```python
print("\n========== FINAL ANALYSIS SUMMARY ==========")

print("Total Employees:", len(data))

print("\nEmployees by Department:")
print(data["Department"].value_counts())

print("\nOverall Average Performance:",
      round(data["Performance_Score"].mean(), 2))

print("Overall Average Salary:",
      round(data["Salary(LPA)"].mean(), 2), "LPA")

print("Overall Average Attendance:",
      round(data["Attendance_%"].mean(), 2), "%")

print("\nDepartment with Highest Average Performance:")
print(department_performance.idxmax())

print("Department with Lowest Average Performance:")
print(department_performance.idxmin())

print("\nCorrelation with Performance Score:")
print(
    data[numerical_cols]
    .corr()["Performance_Score"]
    .sort_values(ascending=False)
    .round(3)
)

data[numerical_cols].describe().to_csv("employee_analysis_summary.csv")
#Save the final numerical summary in csv format
```
